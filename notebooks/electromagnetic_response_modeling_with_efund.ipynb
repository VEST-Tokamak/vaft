# Electromagnetic Response Modeling with EFUND

## Overview

This notebook is a first-draft shell for documenting electromagnetic response modeling with EFUND in the VAFT pipeline. It covers magnetic diagnostics, wall geometry, PF active and PF passive geometry, material resistivity, Green's function-based response calculation, electromagnetic coupling matrix construction, and comparison with EFUND module and table structure.

## Objectives

- Organize geometry inputs for magnetic diagnostics, wall structures, PF active coils, and PF passive conductors.
- Document material resistivity and circuit parameter requirements.
- Outline Green's function-based response calculations.
- Define electromagnetic coupling matrix construction and validation against EFUND outputs.

## Expected Inputs

- Magnetic diagnostics geometry and channel metadata.
- Wall, PF active coil, and PF passive conductor geometry definitions.
- Material resistivity, resistance, inductance, and mutual-coupling parameters.
- EFUND module outputs, database tables, or reference matrix structures.

## Expected Outputs

- Green's function response products for diagnostics and conductors.
- Electromagnetic coupling matrices with documented indexing and units.
- Comparison tables between VAFT calculations and EFUND module outputs.
- Open compatibility notes for EFUND table structure and pipeline integration.

## Requirements to run this page

This notebook is still a shell. It is not unimplemented for lack of a VAFT API:
it needs an external code that VAFT does not ship and cannot install for you.
What follows is what it takes to make it runnable, so the gap is a task rather
than a mystery.

### 1. Obtain EFUND

EFUND is not part of VAFT and cannot be installed from this repository. It is
obtained from its maintainers under their own terms and built locally; the build
story is tracked in
[#226](https://github.com/VEST-Tokamak/vaft/issues/226), which `install/README.md`
points to for exactly this reason. The VAFT installation scripts deliberately
cover the Python environment only.

EFUND is the Green's-function table generator that ships inside an EFIT build rather than separately, so obtaining it means obtaining EFIT. `vaft/data/efit/efund_run_command.sh` is the invocation VAFT expects, and it asks for `$EFUND` to point at the executable in your own build; `vaft/data/efit/mhdin.dat` is the VEST machine description it reads.

### 2. Configure the root

```bash
export EFUND=/path/to/installation
```

The adapters resolve executables beneath that root:

  - `the efund executable itself, not a root directory`

`docs/_guide/Installation.md` section 4 documents this layout for every
supported code.

### 3. What VAFT already does without the binary

The electromagnetic response is available natively, which is why this page is a
comparison rather than the only route to the matrices:

- `vaft.process.electromagnetics` computes the response matrices, the mutual
  couplings and the impedance matrices directly
  (`compute_response_matrix`, `compute_mutual_passive_active`,
  `compute_impedance_matrices`);
- those matrices live in the `em_coupling` IDS, and the packaged shots 41524 and
  41672 carry them;
- `eddy_current_calculation_and_startup_analysis.ipynb` uses them end to end
  today, with no external code at all.

The purpose of this page is therefore verification -- VAFT's own coupling
against EFUND's tables -- not capability.

### 4. Check readiness before running anything

`notebooks/initialize_external_fusion_codes.ipynb` reports every configured root
and names the executables it could not find.
`vaft.code._executables.missing_home_message` produces the same message the
adapters raise, so a failure says which variable to set and which file was
expected rather than failing inside a solver.

### 5. What this page will show once EFUND is available

Its own Expected Outputs, above: EFUND's Green's-function tables for the VEST
machine description, compared element by element against the matrices
`vaft.process.electromagnetics` derives, with the differences attributed.
