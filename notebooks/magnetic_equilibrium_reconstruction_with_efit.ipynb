# Magnetic Equilibrium Reconstruction with EFIT

## Overview

This notebook is a first-draft shell for documenting EFIT-based magnetic equilibrium reconstruction in the VAFT pipeline. It focuses on magnetic diagnostics, eddy-current inputs, PF coil information, and physical constraints required for equilibrium reconstruction.

## Objectives

- Define the EFIT reconstruction inputs produced by upstream VAFT stages.
- Document magnetic diagnostics, eddy-current estimates, and PF coil current requirements.
- Organize physical constraints and reconstruction configuration parameters.
- Specify expected EFIT outputs for downstream equilibrium and stability analysis.

## Expected Inputs

- Processed magnetic diagnostics with calibration metadata.
- PF active coil currents and PF passive eddy-current estimates.
- Plasma current, pressure constraints, and boundary or limiter information.
- EFIT configuration files, time slices, and physical constraint definitions.

## Expected Outputs

- EFIT reconstruction results for selected shots and time slices.
- Equilibrium files, flux maps, boundary descriptions, and residual summaries.
- Diagnostics-fit quality metrics and constraint consistency checks.
- Inputs for MHD equilibrium interpretation and stability workflows.

## Pipeline Context

This notebook documents the equilibrium reconstruction stage of the Snakemake-based VAFT pipeline. It should eventually consume processed diagnostics and eddy-current products, then produce versioned EFIT equilibria for MHD analysis, DCON, RDCON, and GPEC workflows.

## Requirements to run this page

This notebook is still a shell. It is not unimplemented for lack of a VAFT API:
it needs an external code that VAFT does not ship and cannot install for you.
What follows is what it takes to make it runnable, so the gap is a task rather
than a mystery.

### 1. Obtain EFIT

EFIT is not part of VAFT and cannot be installed from this repository. It is
obtained from its maintainers under their own terms and built locally; the build
story is tracked in
[#226](https://github.com/VEST-Tokamak/vaft/issues/226), which `install/README.md`
points to for exactly this reason. The VAFT installation scripts deliberately
cover the Python environment only.

EFIT in particular is distributed under access terms set by its maintainers, not by VAFT: obtaining a copy is a conversation with them, and no amount of configuration here substitutes for it.

### 2. Configure the root

```bash
export EFITHOME=/path/to/installation
```

The adapters resolve executables beneath that root:

  - `bin/efit`

`docs/_guide/Installation.md` section 4 documents this layout for every
supported code.

### 3. What VAFT already does without the binary

`vaft.code.efit` is the input and output side of the workflow, and it works with
no EFIT installed:

- `generate_constraints_ods` builds the magnetic constraint set from an ODS,
  applying channel validity so condemned probes are excluded rather than fitted;
- `generate_kfile` writes the k-file EFIT reads;
- `EFITScientificConfig` and `efit_parameter_grid` describe and sweep the run
  settings;
- the packaged `vaft/data/efit/` g-files let every downstream page -- including
  `mhd_equilibrium_analysis.ipynb` -- work from real EFIT output offline.

So the reconstruction stage can be prepared, inspected and validated here; only
the solve itself is blocked.

### 4. Check readiness before running anything

`notebooks/initialize_external_fusion_codes.ipynb` reports every configured root
and names the executables it could not find.
`vaft.code._executables.missing_home_message` produces the same message the
adapters raise, so a failure says which variable to set and which file was
expected rather than failing inside a solver.

### 5. What this page will show once EFIT is available

Its own Expected Outputs, above: a constrained reconstruction for a VEST shot,
the fit quality against the magnetic measurements, and the equilibrium handed to
`mhd_equilibrium_analysis.ipynb`.
